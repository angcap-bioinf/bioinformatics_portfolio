Task: 

Hai un DataFrame con le colonne Gene, Espressione_Controllo e Espressione_Trattato. 

Devi creare una nuova colonna chiamata Status_Regolazione.

Usa apply impostando axis='columns' (o axis=1) per passare l'intera riga a una funzione personalizzata. La funzione dovrà valutare le due colonne simultaneamente:

- Se Espressione_Trattato è almeno il doppio di Espressione_Controllo, restituisci la stringa "Up-regolato".

- Se è la metà (o meno), restituisci "Down-regolato".

- In tutti gli altri casi, restituisci "Stabile".

```python
dati = {
    'Gene': ['BRCA1', 'TP53', 'MYC', 'EGFR', 'KRAS'],
    'Espressione_Controllo':[10.5, 45.2, 5.0, 100.0, 20.0],
    'Espressione_Trattato':[11.0, 15.1, 15.0, 105.0, 8.5]
}

df= pd.DataFrame(dati)


def diff(riga):

    tratt=riga["Espressione_Trattato"]
    contr=riga["Espressione_Controllo"]


    if tratt>=(contr*2):

        return "Up-regolato"


    elif tratt<=(contr/2):

        return "Down-regolato"

    else:

        return "Stabile"






df["Status_Regolazione"]=df[["Espressione_Controllo","Espressione_Trattato"]].apply(diff, axis="columns")





display(df)
```

```python
dati = {
    'Gene': ['BRCA1', 'TP53', 'MYC', 'EGFR', 'KRAS'],
    'Espressione_Controllo':[10.5, 45.2, 5.0, 100.0, 20.0],
    'Espressione_Trattato':[11.0, 15.1, 15.0, 105.0, 8.5]
}

df= pd.DataFrame(dati)


def diff(tratt,contr):
    out=[]

    for riga in range(len(tratt)):

        if tratt.iloc[riga]>=(contr.iloc[riga]*2):

            out.append("Up-regolato")


        elif tratt.iloc[riga]<=(contr.iloc[riga]/2):

             out.append("Down-regolato")

        else:

             out.append("Stabile")
    
    return(out)





df["Status_Regolazione"]=diff(df["Espressione_Controllo"], df["Espressione_Trattato"])





display(df)
```
