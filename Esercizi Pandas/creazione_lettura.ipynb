```python
import pandas as pd

# --- PASSO 1: Creazione Manuale del DataFrame ---
# Definiamo i dati manualmente in un dizionario per chiarezza
dati = {
    'Gene': ['BRCA1', 'TP53', 'EGFR'],
    'Livello_Espressione': [15.4, 2.1, 8.9],
    'Cromosoma': ['chr17', 'chr17', 'chr7'] # Dati realistici per questi geni
}

# Creiamo il DataFrame
df = pd.DataFrame(dati)



# --- PASSO 2: Salvataggio come CSV ---
# Salviamo il file. Usiamo index=False per non salvare l'indice di pandas (che è 0,1,2...)
# nel file CSV, così quando lo rileggiamo possiamo decidere quale colonna usare come indice.
nome_file = 'dati_geni.csv'
df.to_csv(nome_file, index=False)



# --- PASSO 3: Lettura del CSV con index_col ---
# Rileggiamo il file specificando che la colonna 'Gene' deve diventare l'indice.
pd.read_csv(nome_file, index_col=0)


```
