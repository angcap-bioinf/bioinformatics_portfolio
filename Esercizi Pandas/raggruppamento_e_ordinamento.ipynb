4. Raggruppamento e Ordinamento (groupby, sort_values)

Task: Hai un dataset RNA-seq di pazienti sani e malati. Usa groupby sulla colonna Condition (Healthy/Tumor) per calcolare la media (mean()) dell'espressione del gene "MYC". Usa sort_values in ordine decrescente (ascending=False) per trovare i 5 pazienti con l'espressione più alta in assoluto.

```python
data = {
    'Patient_ID': ['P01', 'P02', 'P03', 'P04', 'P05', 'P06', 'P07', 'P08'],
    'Condition': ['Healthy', 'Healthy', 'Healthy', 'Tumor', 'Tumor', 'Healthy', 'Healthy', 'Tumor'],
    'MYC': [1.2, 15.4, 0.9, 12.1, 1.5, 18.2, 0.8, 14.7]
}

df = pd.DataFrame(data)

display(df)
```

```python
data = {
    'Patient_ID': ['P01', 'P02', 'P03', 'P04', 'P05', 'P06', 'P07', 'P08'],
    'Condition': ['Healthy', 'Healthy', 'Healthy', 'Tumor', 'Tumor', 'Healthy', 'Healthy', 'Tumor'],
    'MYC': [1.2, 15.4, 0.9, 12.1, 1.5, 18.2, 0.8, 14.7]
}

df = pd.DataFrame(data)

media_MYC=df.groupby("Condition").MYC.mean()
top_pat=df.groupby("Patient_ID").MYC.apply(lambda p: max(p)).sort_values(ascending=False).head(5)

display(media_MYC,top_pat)
```
