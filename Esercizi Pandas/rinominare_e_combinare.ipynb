6. Rinominare e Combinare (rename, concat, join)

Task: Hai due file CSV: uno con i metadati dei pazienti (Patient_ID, Age, Stage) e uno con l'espressione genica (Patient_ID, TP53_expr, BRCA1_expr). Uniscili usando un join (o merge) sulla colonna Patient_ID. Usa rename per cambiare Stage in Tumor_Stage.


```python

df_clinico = pd.DataFrame({
    'Patient_ID': ['P001', 'P002', 'P003', 'P004'],
    'Age': [45, 62, 38, 55],
    'Stage': ['I', 'III', 'II', 'IV']
})

df_genica = pd.DataFrame({
    'Patient_ID': ['P001', 'P002', 'P003', 'P005'], 
    'TP53_expr': [12.5, 8.2, 15.1, 9.4],
    'BRCA1_expr': [5.1, 4.8, 7.2, 6.3]
})

display("DataFrame Clinico:\n", df_clinico)
display("\nDataFrame Espressione Genica:\n", df_genica)
```

```python

clinico = pd.DataFrame({
    'Patient_ID': ['P001', 'P002', 'P003', 'P004'],
    'Age': [45, 62, 38, 55],
    'Stage': ['I', 'III', 'II', 'IV']
})

genica = pd.DataFrame({
    'Patient_ID': ['P001', 'P002', 'P003', 'P005'], 
    'TP53_expr': [12.5, 8.2, 15.1, 9.4],
    'BRCA1_expr': [5.1, 4.8, 7.2, 6.3]
})

clinico=clinico.set_index('Patient_ID')
genica=genica.set_index('Patient_ID')

pd.DataFrame(clinico.join(genica,lsuffix="_clin", rsuffix="_gen", how="outer")).rename(columns={"Stage":"Tumor_Stage"})

```

```python

clinico = pd.DataFrame({
    'Patient_ID': ['P001', 'P002', 'P003', 'P004'],
    'Age': [45, 62, 38, 55],
    'Stage': ['I', 'III', 'II', 'IV']
})

genica = pd.DataFrame({
    'Patient_ID': ['P001', 'P002', 'P003', 'P005'], 
    'TP53_expr': [12.5, 8.2, 15.1, 9.4],
    'BRCA1_expr': [5.1, 4.8, 7.2, 6.3]
})


pd.merge(clinico,genica, how="outer").rename(columns={"Stage":"Tumor_Stage"})

```
