Task: Hai una colonna con sequenze di DNA. Usa apply con una funzione lambda o personalizzata per calcolare la lunghezza di ogni sequenza e creare una nuova colonna Seq_Length. Usa value_counts() su una colonna Consequence (es. missense, frameshift, silent) per vedere quante mutazioni ci sono per tipo.

```python
data = {
    'DNA_Sequence': ['ATGCGT', 'ATGC', 'GGGTTAACCA', 'ATGCGTACGT', 'TAGA', 'ATGCGT'],
    'Consequence': ['missense', 'silent', 'missense', 'frameshift', 'silent', 'missense']
}

df = pd.DataFrame(data)

# display(df)

def lung(string):
    count=0
    for nuc in string:
        count+=1
    return count

df["Seq_Length"]= df["DNA_Sequence"].apply(lung)


mut= df.Consequence.value_counts()

display(df[["Seq_Length","DNA_Sequence"]])
display(mut)
```

```python

dati={

"DNA_Sequence":["ATGCGT","ATGC","GGGTTAACCA","ATGCGTACGT","TAGA","ATGCGT"],
"Consequence":["missense","silent","missense","frameshift","silent","missense"]

    
}

df = pd.DataFrame(dati)

# display(df)

def lung(seq):
    c=0
    while c<len(seq):
        c+=1
    return c

df["lunghezza"]=df.DNA_Sequence.map(lung)

conta=df.Consequence.value_counts()

display(df[["lunghezza","DNA_Sequence"]])
display(conta)
```
