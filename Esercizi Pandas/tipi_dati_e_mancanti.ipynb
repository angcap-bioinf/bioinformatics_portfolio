```python
import pandas as pd
import numpy as np

# 1. Caricamento dati
data = {
    'Age': [25, 30, None, 45, None, 50],
    'Sex': ['M', 'F', 'Unknown', 'M', 'F', 'Unknown']
}

df = pd.DataFrame(data)

# 2. Controllo dei dtypes
print("Dtypes iniziali:")
print(df.dtypes)
print("-" * 30)

# 3. Trasformazione "Unknown" in NaN (Usando np.nan come richiesto)
# Facciamo questo per PRIMO, così il dataset è "pulito" e uniforme
df['Sex'] = df['Sex'].replace("Unknown", np.nan)

# 4. Riempimento dei vuoti nella colonna Age con la mediana
# Calcoliamo la mediana e la applichiamo SOLO alla colonna Age
mediana_age = df['Age'].median()
df['Age'] = df['Age'].fillna(mediana_age)

# Visualizzazione risultato finale
print("DataFrame finale:")
display(df)
```

```python
data = {

    'Age': [25, 30, None, 45, None, 50],

    'Sex': ['M', 'F', 'Unknown', 'M', 'F', 'Unknown']

}


df = pd.DataFrame(data)


display(df.dtypes)


df=df.fillna(df.Age.median()).replace("Unknown", None)

display(df)
```

fare ulteriore versione che modifica l'originale e non ne crea uno nuovo

provare anche con apply o map

```python
data = {

    'Age': [25, 30, None, 45, None, 50],

    'Sex': ['M', 'F', 'Unknown', 'M', 'F', 'Unknown']

}


df = pd.DataFrame(data)

display(df.dtypes)

def sost(sex):
    if sex=="Unknown":
        return(None)
    return(sex)


df["Age"]=df.Age.apply(lambda p: df.Age.median() if pd.isna(p) else p)
df["Sex"]=df.Sex.apply(sost)

display(df)
```

```python
data = {

    'Age': [25, 30, None, 45, None, 50],

    'Sex': ['M', 'F', 'Unknown', 'M', 'F', 'Unknown']

}


df = pd.DataFrame(data)

display(df.dtypes)


df.Age=df.Age.apply(lambda p: df.Age.median() if pd.isna(p) else p)
df.Sex=df.Sex.map({"Unknown":None, "M":"M", "F":"F"})

display(df)
```
